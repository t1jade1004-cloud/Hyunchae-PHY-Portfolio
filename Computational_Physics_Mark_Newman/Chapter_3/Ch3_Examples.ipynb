```python
import matplotlib.pyplot as plt
from numpy import linspace, sin, cos

x = linspace(0,10,100)
y1 = sin(x)
y2 = cos(x)
plt.plot(x,y1,"k-")
plt.plot(x,y2,"k--")
plt.ylim(-1.2,1.2)
plt.xlabel("x axis")
plt.ylabel("y = sin x or y = cos x")
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

plt.style.use('_mpl-gallery')

# Make data
X = np.arange(-5, 5, 0.25)
Y = np.arange(-5, 5, 0.25)
X, Y = np.meshgrid(X, Y)
R = np.sqrt(X**2 + Y**2)
Z = np.sin(R)

# Plot the surface
fig, ax = plt.subplots(subplot_kw={"projection": "3d"})
ax.plot_surface(X, Y, Z, vmin=Z.min() * 2, cmap="Blues")

ax.set(xticklabels=[],
       yticklabels=[],
       zticklabels=[])

plt.show()
```
