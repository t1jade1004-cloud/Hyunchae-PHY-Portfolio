07.26.2026 Exercise 2.6: Planetary orbits

```python
#l1 = distance of perihelion
#l2 = distance of aphelion

#v1 = velocity of perihelion
#v2 = velocity of aphelion

#l1 * v1 = l2 * v2

from math import sqrt, pi

G = 6.6743e-11
M = 1.9891e30

l1 = float(input("Enter the distance to the Sun at perihelion: "))
v1 = float(input("Enter the velocity to the Sun at perihelion: "))


A = 1
B = -1 * (2*G*M)/(v1*l1)
C = -1 * ((v1**2)-((2*G*M)/l1))

v2 = (-B - sqrt(B**2 - 4*A*C))/(2*A)

l2 = (l1*v1)/v2
a = 0.5 * (l1+l2)
b = sqrt(l1*l2)
t = (2 * pi * a * b)/(l1*v1)
t_year = t/(60*60*24*365.25)
e = (l2-l1)/(l2+l1)

print(f"Distance at aphelion: {round(l2,2)}")
print(f"Velocity at aphelion: {round(v2,2)}")
print(f"Orbital Period: {round(t_year,0)} years")
print(f"Orbital Eccentricity {round(e,2)}")
```

08.04.2026 Exercise 2.10: Binomial coefficients

```python
def factorial(n):
    result = 1
    for i in range(1,n+1):
        result *= i

    return result

def bcoe(n,k):
    if k == 0:
        return 1
    return int(factorial(n)/(factorial(k)*factorial(n-k)))

l = int(input("Enter the number of lines in Pascal's triangle: "))
for n in range(1,l+1):
    for k in range(0,n+1):
        print(bcoe(n,k),end=" ")
    print("\n")
```

08.04.2026 Exercise 2.11: Prime numbers

```python
# A number n is prime if it has no prime factors less than n. Hence, we need to only check if it is divisible by other primes.
# Any non-prime must have factors, and hence also prime-factors, less than or equal to √n.
# If we find even a signle prime factor less than √n then we know that the number is non-prime, and hence there is no need to check any further.

from math import sqrt

primes = [2]

def checkprime(n):
    for p in primes:
        if p <= sqrt(n):
            if n%p == 0:
                return
            
        else:
            primes.append(n)
            print(n)
            return
        



n = int(input("Enter the value of n: "))

for i in range(3,n+1):
    checkprime(i)

```

08.04.2026 Exercise 2.12: Greatest common divisor

```python
def gcd(m,n):
    if n == 0:
        return m
    elif n >0:
        return gcd(n,m%n)

print(gcd(108, 192))
```
